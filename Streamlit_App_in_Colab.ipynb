```
%%writefile app.py
import streamlit as st
import pandas as pd
import sqlite3
from sqlalchemy import create_engine

# Use SQLite instead of MySQL
engine_imdb = create_engine('sqlite:///movies.db')

PAGE_CONFIG = {"page_title":"SQLite Example","page_icon":":smiley:","layout":"centered"}
st.set_page_config(**PAGE_CONFIG)

def hist_all_movies(sql, engine):
    df_movies = pd.read_sql_query(sql, con=engine)
    if not df_movies.empty:
        fig = df_movies.rating.hist().get_figure()
        st.pyplot(fig)
    else:
        st.write("No data found for this query.")

def main():
    st.title("Streamlit with SQLite")
    st.subheader("Running from Colab using local DB")

    menu = ["Page 1","Page 2", "Page 3"]
    choice = st.sidebar.selectbox('Menu',menu)

    if choice == 'Page 1':
        st.subheader("Welcome")
        st.write("This app now uses a local SQLite database created in your Colab environment.")

    elif choice == 'Page 2':
        SQL_script = st.text_area(label='SQL Input', value='SELECT rating FROM movies')
        hist_all_movies(SQL_script, engine_imdb)

    elif choice == 'Page 3':
        slider_value = st.slider('Number of rows', min_value=1, max_value=10)
        df_movies = pd.read_sql(f"SELECT * FROM movies LIMIT {slider_value}", con=engine_imdb)

        st.write("Movie Ratings Plot:")
        if not df_movies.empty and 'rating' in df_movies.columns:
            fig = df_movies['rating'].plot(kind='bar').get_figure()
            st.pyplot(fig)

        show_dataframe = st.checkbox('Display Dataframe')
        if show_dataframe:
            st.dataframe(df_movies)

if __name__ == '__main__':
    main()
```
